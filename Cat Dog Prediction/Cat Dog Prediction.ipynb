```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns



import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.layers import Conv2D,Dense,Dropout,MaxPooling2D,Flatten
from tensorflow.keras.models import Sequential
```

Generator = All image can not be fit in ram at once so generator send data in Batch and It helps to generate data to

Generator 
1. Load data in chunks
2. Preprocess it
3. send dta to model
4. Repeat Loop

```python
train_ds = keras.utils.image_dataset_from_directory(
    directory='Dataset/dogs_vs_cats/train',
    batch_size=64,
    label_mode='int',   #give integer value to then cat and dog 0 or 1
    labels='inferred',
    image_size=(256,256) #making all img zize 256,256
)


test_ds = keras.utils.image_dataset_from_directory(
    directory='Dataset/dogs_vs_cats/test',
    batch_size=64,
    label_mode='int',   #give integer value to then cat and dog 0 or 1
    labels='inferred',
    image_size=(256,256) #making all img zize 256,256
)


```

TAKING SMALL CHUNK OF DATA CAUSE LONG DATA TAKE LONG TIME

```python
small_train = train_ds.take(100)
small_test = train_ds.take(100)
```

NORMALIZING THE DATA RGB 255 TO - small value

```python
def process_img(image, label):
    image = tf.cast(image, tf.float32) / 255.0
    label = tf.cast(label, tf.float32)
    return image, label


small_train = small_train.map(process_img)
small_test = small_test.map(process_img)
```

Model

```python
model= Sequential()

model.add(Conv2D(32,kernel_size=(3,3),padding='valid',activation='relu',input_shape=(256,256,3)))
model.add(MaxPooling2D(pool_size=(2,2),strides=2,padding='valid'))


model.add(Conv2D(64,kernel_size=(3,3),padding='valid',activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2),strides=2,padding='valid'))


model.add(Conv2D(128,kernel_size=(3,3),padding='valid',activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2),strides=2,padding='valid'))

model.add(Flatten())

model.add(Dense(128,activation='relu'))
model.add(Dense(64,activation='relu'))
model.add(Dense(1,activation='sigmoid'))
          
```

```python
model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
h1 = model.fit(small_train,epochs=10,validation_data=small_test)
```

```python
plt.title("Testing Model Performance")
plt.plot(h1.history['loss'],label = "Train Loss")
plt.plot(h1.history['val_loss'],label = "Validation Loss")
plt.legend()
plt.show()
```

```python
plt.title("Reading Accuracy Data")
plt.plot(h1.history['accuracy'],label = "Train Accuracy")
plt.plot(h1.history['val_accuracy'],label = 'VAlidation Accuracy')
plt.legend()
plt.show()
```

```python
y_sigmoid= model.predict(small_test)   #sigmoid give probability of class 1
```

```python
y_pred = np.where(y_sigmoid > 0.5,1,0)
y_pred[:10]
```
